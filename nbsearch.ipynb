```python
# Source - https://stackoverflow.com/a/66857645
# Posted by Tom F
# Retrieved 2026-07-13, License - CC BY-SA 4.0

import glob    # sjc add

pattern = '../**/*.ipynb'  # '../**/*.ipynb'
query = input('search term: ')

for filepath in glob.iglob(pattern, recursive=True):
    with open(filepath, encoding='utf-8') as file:     # sjc mod
        s = file.read()
        if (s.find(query) > -1):
            print(filepath)
```
